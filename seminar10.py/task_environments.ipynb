# Задание: venv и conda

venv - это встроенный модуль питона для создания виртуальных окружений. 
conda - это менеджер пакетов, он умеет ставить разные библиотеки и создавать окружения.

команды:

Создание окружений (для второго аналогично):
python3 -m venv bioenv
source bioenv/bin/activate
pip install ipykernel jupyterlab numpy pandas matplotlib
deactivate

потом
source ponchik/bin/activate
python -m ipykernel install --user --name ponchik --display-name "Ponchik"
deactivate

source bioenv/bin/activate
python -m ipykernel install --user --name bioenv --display-name "Bioinformatics"
deactivate

Запуск самого Jupyter:
source ponchik/bin/activate
jupyter lab
